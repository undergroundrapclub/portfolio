## Javascript - Uploading an image
Most social platforms allow people to upload images to express themselves. That's one of the key features we'd like to focus on with this project.

By including an 'upload' button while drafting a post, the client may attach an image to their post.

```python
%%js

// CODE_RUNNER: Reaction picker


console.log("Starting reaction picker...");

setTimeout(function() {

    // --------------------------------
    // Remove an old reaction picker
    // --------------------------------

    const oldBox = document.getElementById("reaction-picker");

    if (oldBox) {
        oldBox.remove();
    }


    // --------------------------------
    // Main box
    // --------------------------------

    const box = document.createElement("div");

    box.id = "reaction-picker";

    box.style.position = "fixed";
    box.style.top = "50px";
    box.style.right = "50px";

    box.style.width = "500px";
    box.style.backgroundColor = "white";

    box.style.padding = "25px";

    box.style.border = "3px solid #333";
    box.style.borderRadius = "15px";

    box.style.boxShadow = "0 5px 20px rgba(0,0,0,0.25)";

    box.style.zIndex = "999999";

    box.style.textAlign = "center";
    box.style.fontFamily = "Arial, sans-serif";


    // --------------------------------
    // Title
    // --------------------------------

    const title = document.createElement("h2");

    title.textContent = "Choose your reaction";

    box.appendChild(title);


    // --------------------------------
    // Reaction images
    // --------------------------------

    const laughing = document.createElement("img");

    laughing.src =
        "/portfolio/assets/reactions/laughing.png";

    laughing.width = 90;

    laughing.style.margin = "8px";
    laughing.style.padding = "5px";
    laughing.style.cursor = "pointer";
    laughing.style.border = "3px solid transparent";
    laughing.style.borderRadius = "10px";


    const angry = document.createElement("img");

    angry.src =
        "/portfolio/assets/reactions/angry.png";

    angry.width = 90;

    angry.style.margin = "8px";
    angry.style.padding = "5px";
    angry.style.cursor = "pointer";
    angry.style.border = "3px solid transparent";
    angry.style.borderRadius = "10px";


    const shocked = document.createElement("img");

    shocked.src =
        "/portfolio/assets/reactions/shocked.png";

    shocked.width = 90;

    shocked.style.margin = "8px";
    shocked.style.padding = "5px";
    shocked.style.cursor = "pointer";
    shocked.style.border = "3px solid transparent";
    shocked.style.borderRadius = "10px";


    const sad = document.createElement("img");

    sad.src =
        "/portfolio/assets/reactions/sad.png";

    sad.width = 90;

    sad.style.margin = "8px";
    sad.style.padding = "5px";
    sad.style.cursor = "pointer";
    sad.style.border = "3px solid transparent";
    sad.style.borderRadius = "10px";


    box.appendChild(laughing);
    box.appendChild(angry);
    box.appendChild(shocked);
    box.appendChild(sad);


    // --------------------------------
    // Selected reaction
    // --------------------------------

    let selectedImage = null;


    function selectReaction(image) {

        selectedImage = image.src;

        laughing.style.border =
            "3px solid transparent";

        angry.style.border =
            "3px solid transparent";

        shocked.style.border =
            "3px solid transparent";

        sad.style.border =
            "3px solid transparent";

        image.style.border =
            "3px solid #2196F3";

        console.log("Reaction selected!");
    }


    laughing.onclick = function() {
        selectReaction(laughing);
    };

    angry.onclick = function() {
        selectReaction(angry);
    };

    shocked.onclick = function() {
        selectReaction(shocked);
    };

    sad.onclick = function() {
        selectReaction(sad);
    };


    // --------------------------------
    // Post button
    // --------------------------------

    const postButton = document.createElement("button");

    postButton.textContent = "POST REACTION";

    postButton.style.display = "block";

    postButton.style.margin =
        "20px auto";

    postButton.style.padding =
        "10px 25px";

    postButton.style.fontSize = "16px";

    postButton.style.fontWeight = "bold";

    postButton.style.cursor = "pointer";

    postButton.style.borderRadius = "8px";

    postButton.style.border = "none";


    box.appendChild(postButton);


    // --------------------------------
    // Post area
    // --------------------------------

    const postTitle = document.createElement("h3");

    postTitle.textContent = "Your post:";

    box.appendChild(postTitle);


    const postArea = document.createElement("div");

    postArea.style.minHeight = "120px";

    postArea.style.padding = "15px";

    postArea.style.border =
        "2px dashed #aaa";

    postArea.style.borderRadius = "10px";

    box.appendChild(postArea);


    // --------------------------------
    // Post reaction
    // --------------------------------

    postButton.onclick = function() {

        if (selectedImage === null) {

            alert(
                "Please choose a reaction first!"
            );

            return;
        }


        postArea.innerHTML = "";


        const postedImage =
            document.createElement("img");

        postedImage.src =
            selectedImage;

        postedImage.width = 200;


        postArea.appendChild(postedImage);


        console.log("Reaction posted!");
    };


    // --------------------------------
    // Add picker to page
    // --------------------------------

    document.body.appendChild(box);

    console.log("Reaction picker created!");

}, 100);
```

## Python: Uploading text

Text is one of the main ways users communicate and share their thoughts on social media. For this project, we'd like to focus on allowing users to create text-based posts.

By providing a text input, the client can type a message and add it to their post.


```python
# CODE_RUNNER: Create a text post.

posts = []

post = input("Write your post: ")

posts.append(post)

print("\nPosts:")
for post in posts:
    print("- " + post)
```
